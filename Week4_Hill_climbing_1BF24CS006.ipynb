```python

import random

N = 4


def calculate_cost(state):
    """Count pairs of queens attacking each other."""
    cost = 0

    for i in range(N):
        for j in range(i + 1, N):
            # Same row or same diagonal
            if state[i] == state[j] or abs(state[i] - state[j]) == abs(i - j):
                cost += 1

    return cost


def best_successor(current):
    """Generate neighbors by swapping row positions."""
    best_state = current[:]
    best_cost = calculate_cost(current)

    # Try swapping every pair of queens
    for i in range(N):
        for j in range(i + 1, N):
            neighbor = current[:]

            # Swap row positions of two queens
            neighbor[i], neighbor[j] = neighbor[j], neighbor[i]

            cost = calculate_cost(neighbor)

            print("Neighbor:", neighbor, "Cost:", cost)

            if cost < best_cost:
                best_state = neighbor
                best_cost = cost

    return best_state


def hill_climbing():
    # Random initial state: one queen per column
    current = random.sample(range(N), N)

    print("Initial state:", current)
    print("Initial cost:", calculate_cost(current))

    while True:
        print("\nCurrent state:", current)
        current_cost = calculate_cost(current)

        neighbor = best_successor(current)
        neighbor_cost = calculate_cost(neighbor)

        # Stop if no improvement is possible
        if neighbor_cost >= current_cost:
            break

        current = neighbor
        print("Selected state:", current)
        print("Selected cost:", neighbor_cost)

    if calculate_cost(current) == 0:
        print("\nSolution found!")
        print("Final state:", current)
        print("Final cost:", calculate_cost(current))
        print_board(current)
    else:
        print("\nLocal minimum reached.")
        print("Final state:", current)
        print("Final cost:", calculate_cost(current))
        print("Try running the program again.")


def print_board(state):
    """Display the chessboard."""
    for row in range(N):
        for col in range(N):
            if state[col] == row:
                print("Q", end=" ")
            else:
                print(".", end=" ")
        print()


hill_climbing()
```
