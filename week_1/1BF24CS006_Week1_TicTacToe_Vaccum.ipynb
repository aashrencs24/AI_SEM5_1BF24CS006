```python
board = [" " for _ in range(9)]

def print_board():
    print()
    print(f" {board[0]} | {board[1]} | {board[2]} ")
    print("---+---+---")
    print(f" {board[3]} | {board[4]} | {board[5]} ")
    print("---+---+---")
    print(f" {board[6]} | {board[7]} | {board[8]} ")
    print()

def check_winner(player):
    winning_positions = [
        (0, 1, 2),
        (3, 4, 5),
        (6, 7, 8),
        (0, 3, 6),
        (1, 4, 7),
        (2, 5, 8),
        (0, 4, 8),
        (2, 4, 6)
    ]

    for a, b, c in winning_positions:
        if board[a] == board[b] == board[c] == player:
            return True

    return False

def check_draw():
    return " " not in board

print("Welcome to Tic-Tac-Toe!")
print("Player 1 = X")
print("Player 2 = O")
print("Choose a position from 1 to 9.")

current_player = "X"

while True:
    print_board()

    try:
        position = int(input(f"Player {current_player}, choose a position (1-9): "))

        if position < 1 or position > 9:
            print("Please enter a number between 1 and 9.")
            continue

        position -= 1

        if board[position] != " ":
            print("That position is already occupied. Try again.")
            continue

        board[position] = current_player

        if check_winner(current_player):
            print_board()
            print(f" Player {current_player} wins!")
            break


        if check_draw():
            print_board()
            print("It's a draw!")
            break

        current_player = "O" if current_player == "X" else "X"

    except ValueError:
        print("Invalid input. Please enter a number from 1 to 9.")
```

```python
class VacuumCleaner:
    def __init__(self, location, rooms):
        self.location = location
        self.rooms = rooms

    def perceive_and_act(self):

        if self.rooms[self.location] == "Dirty":
            print(f"Vacuum is in Room {self.location} - Room is Dirty")
            print("Action: SUCK")
            self.rooms[self.location] = "Clean"

        else:
            print(f"Vacuum is in Room {self.location} - Room is Clean")

            if self.location == "A":
                print("Action: MOVE RIGHT (A -> B)")
                self.location = "B"
            else:
                print("Action: MOVE LEFT (B -> A)")
                self.location = "A"

        print("Current state:", self.rooms)
        print("-" * 40)

rooms = {
    "A": "Dirty",
    "B": "Dirty"
}

vacuum = VacuumCleaner("A", rooms)

for i in range(6):
    vacuum.perceive_and_act()
```

```python
n = int(input("Enter number of rooms: "))

rooms = []

# Input room status
for i in range(n):
    status = int(input(f"Enter status of Room {i + 1} (1=Dirty, 0=Clean): "))
    rooms.append(status)

print("\nInitial Room Status:", rooms)

# Goal: All rooms must be clean
goal = [0] * n

print("\n--- Vacuum Cleaner Agent ---")

# Visit each room
for i in range(n):

    print(f"\nVacuum is in Room {i + 1}")

    # Simple Reflex Agent
    if rooms[i] == 1:
        print("Room is Dirty -> Clean the room")
        rooms[i] = 0
    else:
        print("Room is Clean -> Move to next room")

    # Check Goal
    if rooms == goal:
        print("\nGoal Achieved: All rooms are clean!")
        break

# Final state
print("\nFinal Room Status:", rooms)

if rooms == goal:
    print(" Goal achieved successfully.")
else:
    print("Goal-Based Agent: Goal not achieved.")
```
