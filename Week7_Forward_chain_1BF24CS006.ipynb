```python
# Forward Chaining in First Order Logic with Dynamic Inputs

# Step 1: Initialize Knowledge Base with user input
print("--- Input Initial Knowledge Base Facts ---")
print("Enter facts in format: Relation,Arg1,Arg2... (e.g., American,Robert or Enemy,A,America)")
print("Enter 'done' when you are finished entering facts.")

KB = []
while True:
    user_input = input("Enter a fact (or 'done'): ").strip()
    if user_input.lower() == 'done':
        break
    if user_input:
        fact_parts = tuple(part.strip() for part in user_input.split(','))
        KB.append(fact_parts)

# Step 2: Define the goal via user input
print("\n--- Input the Goal/Query ---")
goal_input = input("Enter the goal fact (e.g., Criminal,Robert): ").strip()
GOAL = tuple(part.strip() for part in goal_input.split(','))

print("\nInitial Knowledge Base:")
for fact in KB:
    print(fact)
print(f"Goal: {GOAL}")

# Step 3: Forward Chaining
while True:
    NEW_FACTS = []

    # Rule 1: Missile(x) -> Weapon(x)
    for fact in KB:
        if fact[0] == "Missile":
            new_fact = ("Weapon", fact[1])
            if new_fact not in KB and new_fact not in NEW_FACTS:
                NEW_FACTS.append(new_fact)

    # Rule 2: Missile(x) AND Owns(A,x) -> Sells(Robert,x,A)
    for fact in KB:
        if fact[0] == "Missile":
            x = fact[1]
            if ("Owns", "A", x) in KB:
                new_fact = ("Sells", "Robert", x, "A")
                if new_fact not in KB and new_fact not in NEW_FACTS:
                    NEW_FACTS.append(new_fact)

    # Rule 3: Enemy(x,America) -> Hostile(x)
    for fact in KB:
        if fact[0] == "Enemy" and fact[2] == "America":
            new_fact = ("Hostile", fact[1])
            if new_fact not in KB and new_fact not in NEW_FACTS:
                NEW_FACTS.append(new_fact)

    # Rule 4: American(p) AND Weapon(q) AND Sells(p,q,r) AND Hostile(r) -> Criminal(p)
    for fact in KB:
        if fact[0] == "American":
            p = fact[1]
            for weapon in KB:
                if weapon[0] == "Weapon":
                    q = weapon[1]
                    for sale in KB:
                        if sale[0] == "Sells" and sale[1] == p and sale[2] == q:
                            r = sale[3]
                            if ("Hostile", r) in KB:
                                new_fact = ("Criminal", p)
                                if new_fact not in KB and new_fact not in NEW_FACTS:
                                    NEW_FACTS.append(new_fact)

    # Step 4: Check whether goal is generated
    print("\nNew Facts Generated in this iteration:")
    for fact in NEW_FACTS:
        print(fact)

    if GOAL in NEW_FACTS or GOAL in KB:
        print("\nGOAL PROVED!")
        print(f"{GOAL} is true.")
        break

    # Step 5: If no new facts, stop
    if not NEW_FACTS:
        print("\nGOAL CANNOT BE PROVED!")
        break

    # Step 6: Update Knowledge Base
    KB.extend(NEW_FACTS)
```

```python
# Forward Chaining in First Order Logic with Step-by-Step Tracing (Predefined KB)

# Step 1: Predefined Initial Knowledge Base
KB = [
    ("American", "Robert"),
    ("Enemy", "A", "America"),
    ("Owns", "A", "T1"),
    ("Missile", "T1")
]

# Step 2: Predefined Goal
GOAL = ("Criminal", "Robert")

print("=================== STARTING FORWARD CHAINING ===================")
print(f"Target Goal: {GOAL}")
print("Initial Knowledge Base (KB):")
for fact in KB:
    print(f"  - {fact}")

iteration = 0
# Step 3: Forward Chaining Loop
while True:
    iteration += 1
    print(f"\n--- Iteration {iteration} ---")
    NEW_FACTS = []

    # Rule 1: Missile(x) -> Weapon(x)
    print("Checking Rule 1: Missile(x) -> Weapon(x)")
    for fact in KB:
        if fact[0] == "Missile":
            x = fact[1]
            new_fact = ("Weapon", x)
            if new_fact not in KB and new_fact not in NEW_FACTS:
                NEW_FACTS.append(new_fact)
                print(f"  -> Inferred: {new_fact} (Since Missile({x}) is in KB)")

    # Rule 2: Missile(x) AND Owns(A,x) -> Sells(Robert,x,A)
    print("Checking Rule 2: Missile(x) AND Owns(A,x) -> Sells(Robert,x,A)")
    for fact in KB:
        if fact[0] == "Missile":
            x = fact[1]
            if ("Owns", "A", x) in KB:
                new_fact = ("Sells", "Robert", x, "A")
                if new_fact not in KB and new_fact not in NEW_FACTS:
                    NEW_FACTS.append(new_fact)
                    print(f"  -> Inferred: {new_fact} (Since Missile({x}) and Owns(A,{x}) are in KB)")

    # Rule 3: Enemy(x,America) -> Hostile(x)
    print("Checking Rule 3: Enemy(x,America) -> Hostile(x)")
    for fact in KB:
        if fact[0] == "Enemy" and fact[2] == "America":
            x = fact[1]
            new_fact = ("Hostile", x)
            if new_fact not in KB and new_fact not in NEW_FACTS:
                NEW_FACTS.append(new_fact)
                print(f"  -> Inferred: {new_fact} (Since Enemy({x},America) is in KB)")

    # Rule 4: American(p) AND Weapon(q) AND Sells(p,q,r) AND Hostile(r) -> Criminal(p)
    print("Checking Rule 4: American(p) AND Weapon(q) AND Sells(p,q,r) AND Hostile(r) -> Criminal(p)")
    for fact in KB:
        if fact[0] == "American":
            p = fact[1]
            for weapon in KB:
                if weapon[0] == "Weapon":
                    q = weapon[1]
                    for sale in KB:
                        if sale[0] == "Sells" and sale[1] == p and sale[2] == q:
                            r = sale[3]
                            if ("Hostile", r) in KB:
                                new_fact = ("Criminal", p)
                                if new_fact not in KB and new_fact not in NEW_FACTS:
                                    NEW_FACTS.append(new_fact)
                                    print(f"  -> Inferred: {new_fact} (Since American({p}), Weapon({q}), Sells({p},{q},{r}), and Hostile({r}) are in KB)")

    print("\nSummary of inferences for this iteration:")
    if NEW_FACTS:
        for fact in NEW_FACTS:
            print(f"  + New fact added to KB: {fact}")
    else:
        print("  (No new facts inferred in this iteration)")

    # Step 4: Check whether goal is generated
    if GOAL in NEW_FACTS or GOAL in KB:
        print("\n=================== SUCCESS ===================")
        print(f"GOAL PROVED in Iteration {iteration}!")
        print(f"{GOAL} is true.")
        break

    # Step 5: If no new facts, stop to prevent infinite loop
    if not NEW_FACTS:
        print("\n=================== FAILURE ===================")
        print("GOAL CANNOT BE PROVED! (No further derivations possible)")
        break

    # Step 6: Update Knowledge Base
    KB.extend(NEW_FACTS)
    print("Current updated KB state:")
    for fact in KB:
        print(f"  - {fact}")
```
