```python


initial = (
2, 8, 3,
1, 6, 4,
7, 0, 5
)

goal = (
1, 2, 3,
8, 0, 4,
7, 6, 5
)



def print_puzzle(state):
  for i in range(0, 9, 3):
    print(state[i], state[i + 1], state[i + 2])
    print()



def generate_moves(state):

  children = []

  blank = state.index(0)
  row = blank // 3
  col = blank % 3

  if row > 0:
    new_state = list(state)
    new_state[blank], new_state[blank - 3] =  new_state[blank - 3], new_state[blank]
    children.append(("Up", tuple(new_state)))


  if col < 2:
    new_state = list(state)
    new_state[blank], new_state[blank + 1] =  new_state[blank + 1], new_state[blank]
    children.append(("Right", tuple(new_state)))

# Move: Down
  if row < 2:
    new_state = list(state)
    new_state[blank], new_state[blank + 3] =  new_state[blank + 3], new_state[blank]
    children.append(("Down", tuple(new_state)))

# Move: Left
  if col > 0:
    new_state = list(state)
    new_state[blank], new_state[blank - 1] =  new_state[blank - 1], new_state[blank]
    children.append(("Left", tuple(new_state)))
  return children


def depth_limited_search(state, goal, limit, path, moves):


  if state == goal:
    return path, moves

  if limit == 0:
    return None

# Step 3: Generate possible moves
  for move, child in generate_moves(state):
    if len(path) > 0 and child == path[-1]:
      continue

    result = depth_limited_search(
      child,
      goal,
      limit - 1,
      path + [child],
      moves + [move]
    )

    if result is not None:
      return result
  return None


# Iterative Deepening Search
def iterative_deepening_search(initial, goal):

  depth = 0

  while True:

    print("Searching at depth:", depth)

    result = depth_limited_search(
      initial,
      goal,
      depth,
      [initial],
      []
    )

    if result is not None:
      return result

    depth += 1


solution = iterative_deepening_search(initial, goal)

path, moves = solution

print("\nSolution Found!")
print("Number of moves:", len(moves))
print("Moves:", " -> ".join(moves))
print("\nState Space / Solution Path:")

for i, state in enumerate(path):

  print("Step", i)

  if i > 0:
    print("Move:", moves[i - 1])
    print_puzzle(state)
```
